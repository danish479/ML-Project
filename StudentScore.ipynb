```python
# ==========================================
# پروژه پیش‌بینی نمره دانشجویان
# ==========================================

# مرحله ۱: وارد کردن کتابخانه‌های مورد نیاز

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print ("The libraries were imported successfully!!")
```

```python

# ==========================================
# مرحله ۲: ایجاد دیتاست
# ==========================================

np.random.seed(42)

# تعداد دانشجویان
students = 1000

# تولید ساعات مطالعه بین ۱ تا ۱۰ ساعت
study_hours = np.random.uniform(1, 10, students)
# تولید درصد حضور بین ۵۰ تا ۱۰۰ درصد
attendance = np.random.uniform(50, 100, students)
# تولید درصد انجام تکالیف بین ۴۰ تا ۱۰۰ درصد
homework = np.random.uniform(40, 100, students)
```

```python

# ==========================================
# مرحله ۳: محاسبه نمره نهایی
# ==========================================
# محاسبه نمره بر اساس ساعات مطالعه،
# حضور و انجام تکالیف
score = (
    study_hours * 6
    + attendance * 0.20
    + homework * 0.20
)

noise = np.random.normal(0, 5, students)
# محاسبه نمره نهایی
final_score = score + noise
# محدود کردن نمره بین ۰ تا ۱۰۰
final_score = np.clip(final_score, 0, 100)
```

```python

# ==========================================
# مرحله ۴: ایجاد جدول داده‌ها
# ==========================================

data = pd.DataFrame({
    "study_hours": study_hours,
    "attendance": attendance,
    "homework": homework,
    "final_score": final_score
})

```

```python

# ==========================================
# مرحله ۵: بررسی دیتاست
# ==========================================

# نمایش تعداد سطرها و ستون‌ها

print("شکل دیتاست:", data.shape)

# نمایش پنج دانشجوی اول

print("\nپنج دانشجوی اول:")
print(data.head())

# نمایش آمار کلی دیتاست

print("\nآمار دیتاست:")
print(data.describe())
```

```python

# ==========================================
# مرحله ۶: مشخص کردن ویژگی‌ها و نمره هدف
# ==========================================

# ویژگی‌هایی که مدل برای پیش‌بینی استفاده می‌کند

X = data[
    [
        "study_hours",
        "attendance",
        "homework"
    ]
]

# نمره نهایی که می‌خواهیم پیش‌بینی کنیم

y = data["final_score"]
```

```python

# ==========================================
# مرحله ۷: تقسیم دیتاست به آموزش و آزمایش
# ==========================================

# ۸۰ درصد داده‌ها برای آموزش
# ۲۰ درصد داده‌ها برای آزمایش

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nتعداد دانشجویان آموزشی:", len(X_train))
print("تعداد دانشجویان آزمایشی:", len(X_test))

```

```python

# ==========================================
# مرحله ۸: ایجاد مدل رگرسیون خطی
# ==========================================

# ایجاد مدل یادگیری ماشین

model = LinearRegression()

# آموزش مدل با داده‌های آموزشی

model.fit(X_train, y_train)

```

```python

# ==========================================
# مرحله ۹: پیش‌بینی نمرات دانشجویان آزمایشی
# ==========================================

# استفاده از مدل آموزش‌دیده برای پیش‌بینی

y_pred = model.predict(X_test)

# محدود کردن پیش‌بینی‌ها بین ۰ تا ۱۰۰

y_pred = np.clip(y_pred, 0, 100)
```

```python

# ==========================================
# مرحله ۱۰: ارزیابی مدل
# ==========================================

# محاسبه میانگین قدر مطلق خطا

mae = mean_absolute_error(y_test, y_pred)

# محاسبه میانگین مربع خطا

mse = mean_squared_error(y_test, y_pred)

# محاسبه جذر میانگین مربع خطا

rmse = np.sqrt(mse)

# محاسبه ضریب تعیین

r2 = r2_score(y_test, y_pred)

print("\nعملکرد مدل")
print("-----------------")
print("MAE :", round(mae, 2))
print("MSE :", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```python

# ==========================================
# مرحله ۱۱: پیش‌بینی نمره یک دانشجوی جدید
# ==========================================

# اطلاعات دانشجوی جدید

new_student = pd.DataFrame({
    "study_hours": [8],
    "attendance": [95],
    "homework": [90]
})

# پیش‌بینی نمره دانشجوی جدید

prediction = model.predict(new_student)

# محدود کردن نمره پیش‌بینی‌شده بین ۰ تا ۱۰۰

prediction = np.clip(prediction, 0, 100)

print("\nدانشجوی جدید")
print("-----------------")
print("ساعات مطالعه:", new_student["study_hours"].iloc[0])
print("حضور:", new_student["attendance"].iloc[0], "%")
print("تکالیف:", new_student["homework"].iloc[0], "%")
print("نمره پیش‌بینی‌شده:", round(prediction[0], 2))
```
